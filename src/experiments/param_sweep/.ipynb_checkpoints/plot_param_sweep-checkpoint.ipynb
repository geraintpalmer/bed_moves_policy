```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import matplotlib.patches as patches
import matplotlib.patheffects as path_effects
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
gammas_S1 = [0.5, 0.75, 0.9, 0.95, 0.99, 0.995, 0.999, 0.9995]
```

```python
maxcost_S1 = pd.read_csv(f'sweep_gamma_s1/gamma_05000/results/evaluation.csv')['Stage 0'].median()
```

```python
best_cost_S1 = []
for g in gammas_S1:
    gstr = str(g)[2:].ljust(4, '0')
    C = pd.read_csv(f'sweep_gamma_s1/gamma_0{gstr}/results/evaluation.csv').median().min()
    best_cost_S1.append(((maxcost_S1 - C) / maxcost_S1) * 100)
```

```python
fig, ax = plt.subplots(1)
ax.plot(range(len(gammas_S1)), best_cost_S1)
ax.scatter(range(len(gammas_S1)), best_cost_S1)
ax.set_xticks(range(len(gammas_S1)))
ax.set_xticklabels(gammas_S1)
ax.set_ylabel("Cost Reduction after Learning (%)")
ax.set_xlabel(r"Discounting Factor ($\gamma$)")
plt.show()
```

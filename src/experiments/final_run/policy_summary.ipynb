```python
import numpy as np
import tqdm.notebook as tqdm
```

```python
keys = np.memmap('results/stage_31_overall_policykeys_epsilon_1.0.bin', dtype=np.int64, mode='r')
policy = np.memmap('results/stage_31_overall_policyactions_epsilon_1.0.bin', dtype=np.int16, mode='r')
```

```python
summary = {
    '0': {'insert': 0, 'move': 0, 'move_to_surge': 0},
    '1': {'insert': 0, 'move': 0, 'move_to_surge': 0},
    '2': {'insert': 0, 'move': 0, 'move_to_surge': 0}
}
```

```python
for s, a in tqdm.tqdm(zip(keys, policy)):
    state = str(s).zfill(5)
    p = state[-5]
    action = str(a).zfill(4)
    a1 = action[:2]
    a2 = action[2:]
    if a1 == a2:
        summary[p]['insert'] += 1
    else:
        if a2 == '15':
            summary[p]['move_to_surge'] += 1
        else:
            summary[p]['move'] += 1
```
